```python
import numpy as np

def summation(a, b, w0, w1, w2):
    y = w0 + w1 * a + w2 * b
    return y

def step(y):
    if y >= 0:
        return 1
    else:
        return 0

def bipolar_step(y):
    if y >= 0:
        return 1
    else:
        return -1

def sigmoid(y):
    return 1 / (1 + np.exp(-y))

def tanh(y):
    return np.tanh(y)

def relu(y):
    if y >= 0:
        return y
    else:
        return 0

def leaky_relu(y):
    if y >= 0:
        return y
    else:
        return 0.01 * y

def comparator(actual, predicted):
    error = actual - predicted
    return error

a = 1
b = 1

w0 = 10
w1 = 0.2
w2 = -0.75

y = summation(a, b, w0, w1, w2)

print("Summation:", y)
print("Step:", step(y))
print("Bipolar Step:", bipolar_step(y))
print("Sigmoid:", sigmoid(y))
print("TanH:", tanh(y))
print("ReLU:", relu(y))
print("Leaky ReLU:", leaky_relu(y))
print("Error:", comparator(1, step(y)))
```

```python
import numpy as np
import matplotlib.pyplot as plt

def step(y):
    if y >= 0:
        return 1
    else:
        return 0

def perceptron(x, target, learning_rate):

    w0 = 10
    w1 = 0.2
    w2 = -0.75

    errors = []

    for epoch in range(1000):

        total_error = 0

        for i in range(len(x)):

            a = x[i][0]
            b = x[i][1]

            y = w0 + w1 * a + w2 * b

            prediction = step(y)

            error = target[i] - prediction

            w0 = w0 + learning_rate * error
            w1 = w1 + learning_rate * error * a
            w2 = w2 + learning_rate * error * b

            total_error = total_error + error ** 2

        errors.append(total_error)

        if total_error <= 0.002:
            break

    return w0, w1, w2, epoch + 1, errors

x = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

target = np.array([0, 0, 0, 1])

w0, w1, w2, epochs, errors = perceptron(
    x, target, 0.05
)

print("Final W0:", w0)
print("Final W1:", w1)
print("Final W2:", w2)
print("Number of epochs:", epochs)

plt.plot(range(1, len(errors) + 1), errors)
plt.xlabel("Epoch")
plt.ylabel("Sum Square Error")
plt.title("AND Gate - Error vs Epoch")
plt.show()
```

```python
import numpy as np
import os
from PIL import Image
from sklearn.metrics import accuracy_score


def load_images(path):

    images = []
    labels = []

    classes = ["normal", "glaucoma"]

    for class_name in classes:

        folder = os.path.join(path, class_name)

        for file in os.listdir(folder):

            file_path = os.path.join(folder, file)

            try:

                image = Image.open(file_path).convert("L")
                image = image.resize((16, 16))

                image = np.array(image)
                image = image / 255.0

                images.append(image.flatten())

                if class_name == "normal":
                    labels.append(0)
                else:
                    labels.append(1)

            except:
                pass

    return np.array(images), np.array(labels)


def sigmoid(x):
    return 1 / (1 + np.exp(-x))


def train_perceptron(x, y):

    weights = np.zeros(x.shape[1] + 1)

    for epoch in range(100):

        for i in range(len(x)):

            input_value = np.insert(x[i], 0, 1)

            output = np.dot(weights, input_value)

            prediction = sigmoid(output)

            error = y[i] - prediction

            weights = weights + 0.05 * error * input_value

    return weights


def predict(x, weights):

    predictions = []

    for i in range(len(x)):

        input_value = np.insert(x[i], 0, 1)

        output = np.dot(weights, input_value)

        prediction = sigmoid(output)

        if prediction >= 0.5:
            predictions.append(1)
        else:
            predictions.append(0)

    return np.array(predictions)

train_path = "/content/drive/MyDrive/RIM-ONE_DL_images/partitioned_randomly/training_set"
test_path = "/content/drive/MyDrive/RIM-ONE_DL_images/partitioned_randomly/test_set"


# Load data
x_train, y_train = load_images(train_path)
x_test, y_test = load_images(test_path)

print("Training images:", len(x_train))
print("Testing images:", len(x_test))
print("Number of features:", x_train.shape[1])


# Train model
weights = train_perceptron(x_train, y_train)

# Prediction
prediction = predict(x_test, weights)

accuracy = accuracy_score(y_test, prediction)

print(" Accuracy:", accuracy * 100, "%")
```

```python
import numpy as np
from sklearn.metrics import accuracy_score


def pseudo_inverse(x, y):

    x = np.column_stack(
        (np.ones(len(x)), x)
    )

    weights = np.linalg.pinv(x) @ y

    return weights


def predict_pseudo_inverse(x, weights):

    x = np.column_stack(
        (np.ones(len(x)), x)
    )

    output = x @ weights

    prediction = []

    for value in output:

        if value >= 0.5:
            prediction.append(1)
        else:
            prediction.append(0)

    return np.array(prediction)


# Pseudo-inverse
weights_pinv = pseudo_inverse(
    x_train,
    y_train
)

prediction_pinv = predict_pseudo_inverse(
    x_test,
    weights_pinv
)

pinv_accuracy = accuracy_score(
    y_test,
    prediction_pinv
)

print("Perceptron Accuracy:",
      accuracy * 100, "%")

print("Pseudo-Inverse Accuracy:",
      pinv_accuracy * 100, "%")
```

```python
import numpy as np
import matplotlib.pyplot as plt


def sigmoid(x):
    return 1 / (1 + np.exp(-x))


def train_network(x, target):

    np.random.seed(10)

    w1 = np.random.uniform(-0.5, 0.5, (2, 2))
    w2 = np.random.uniform(-0.5, 0.5, 2)

    b1 = np.zeros(2)
    b2 = 0

    errors = []

    for epoch in range(1000):

        total_error = 0

        for i in range(len(x)):

            hidden_input = np.dot(x[i], w1) + b1
            hidden_output = sigmoid(hidden_input)

            final_input = np.dot(hidden_output, w2) + b2
            output = sigmoid(final_input)

            error = target[i] - output

            total_error = total_error + error ** 2

            output_delta = error * output * (1 - output)

            hidden_delta = (
                hidden_output *
                (1 - hidden_output) *
                output_delta * w2
            )

            w2 = w2 + 0.05 * output_delta * hidden_output
            b2 = b2 + 0.05 * output_delta

            w1 = w1 + 0.05 * np.outer(
                x[i],
                hidden_delta
            )

            b1 = b1 + 0.05 * hidden_delta

        errors.append(total_error)

        if total_error <= 0.002:
            break

    return epoch + 1, errors


x = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

target = np.array([0, 0, 0, 1])

epochs, errors = train_network(x, target)

print("Number of epochs:", epochs)

plt.plot(range(1, len(errors) + 1), errors)
plt.xlabel("Epoch")
plt.ylabel("Error")
plt.title("A8 - Backpropagation for AND")
plt.show()
```

```python
import numpy as np


def sigmoid(x):
    return 1 / (1 + np.exp(-x))


def train_two_output(x, target):

    np.random.seed(10)

    w1 = np.random.uniform(-0.5, 0.5, (2, 2))
    w2 = np.random.uniform(-0.5, 0.5, (2, 2))

    b1 = np.zeros(2)
    b2 = np.zeros(2)

    for epoch in range(1000):

        total_error = 0

        for i in range(len(x)):

            hidden_input = np.dot(x[i], w1) + b1
            hidden_output = sigmoid(hidden_input)

            output_input = (
                np.dot(hidden_output, w2) + b2
            )

            output = sigmoid(output_input)

            error = target[i] - output

            total_error = total_error + np.sum(error ** 2)

            output_delta = (
                error * output * (1 - output)
            )

            hidden_error = np.dot(
                output_delta,
                w2.T
            )

            hidden_delta = (
                hidden_output *
                (1 - hidden_output) *
                hidden_error
            )

            w2 = w2 + 0.05 * np.outer(
                hidden_output,
                output_delta
            )

            b2 = b2 + 0.05 * output_delta

            w1 = w1 + 0.05 * np.outer(
                x[i],
                hidden_delta
            )

            b1 = b1 + 0.05 * hidden_delta

        if total_error <= 0.002:
            break

    return epoch + 1


x = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

# AND gate
# 0 -> [1, 0]
# 1 -> [0, 1]

target = np.array([
    [1, 0],
    [1, 0],
    [1, 0],
    [0, 1]
])

epochs = train_two_output(x, target)

print("Number of epochs:", epochs)
```

```python
import numpy as np
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score


# AND gate
and_x = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

and_y = np.array([0, 0, 0, 1])


and_model = MLPClassifier(
    hidden_layer_sizes=(2,),
    max_iter=2000,
    random_state=10
)

and_model.fit(and_x, and_y)

and_prediction = and_model.predict(and_x)

print("AND predictions:", and_prediction)
print("AND accuracy:",
      accuracy_score(and_y, and_prediction) * 100)


# XOR gate
xor_x = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

xor_y = np.array([0, 1, 1, 0])


xor_model = MLPClassifier(
    hidden_layer_sizes=(4,),
    max_iter=5000,
    random_state=10
)

xor_model.fit(xor_x, xor_y)

xor_prediction = xor_model.predict(xor_x)

print("XOR predictions:", xor_prediction)
print("XOR accuracy:",
      accuracy_score(xor_y, xor_prediction) * 100)
```
