```python
import numpy as np
import os

from PIL import Image

from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score
from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report
```

```python
def load_images(path):

    images = []
    labels = []

    classes = ["normal", "glaucoma"]
    for class_name in classes:

        folder = os.path.join(path, class_name)
        for file in os.listdir(folder):
            file_path = os.path.join(folder, file)
            try:
                image = Image.open(file_path).convert("L")
                image = image.resize((16, 16))
                image = np.array(image)
                image = image / 255.0
                images.append(image.flatten())

                if class_name == "normal":
                    labels.append(0)
                else:
                    labels.append(1)
            except:
                pass
    return np.array(images), np.array(labels)
```

```python
train_path = "/content/drive/MyDrive/RIM-ONE_DL_images/partitioned_randomly/training_set"
test_path = "/content/drive/MyDrive/RIM-ONE_DL_images/partitioned_randomly/test_set"
```

```python
x_train, y_train = load_images(train_path)
x_test, y_test = load_images(test_path)
print("Training data:", x_train.shape)
print("Testing data:", x_test.shape)
```

```python
model = MLPClassifier(
    hidden_layer_sizes=(64, 32),
    activation="relu",
    solver="adam",
    max_iter=100,
    random_state=10
)
```

```python
model.fit(x_train, y_train)
```

```python
train_prediction = model.predict(x_train)
test_prediction = model.predict(x_test)
```

```python
train_accuracy = accuracy_score(
    y_train,
    train_prediction
)
test_accuracy = accuracy_score(
    y_test,
    test_prediction
)
print("Training Accuracy:",
      train_accuracy * 100, "%")
print("Testing Accuracy:",
      test_accuracy * 100, "%")
```

```python
cm = confusion_matrix(
    y_test,
    test_prediction
)
print("Confusion Matrix:")
print(cm)
```

```python
print("Classification Report:")
print(
    classification_report(
        y_test,
        test_prediction,
        target_names=["normal", "glaucoma"]
    )
)
```
